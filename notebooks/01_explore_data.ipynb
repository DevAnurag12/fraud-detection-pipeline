## **1. Load both files and count the payments**

```python
import pandas as pd

train = pd.read_csv("../data/fraudTrain.csv")
test = pd.read_csv("../data/fraudTest.csv")

print(len(train), len(test))
```

## **2. How rare is fraud?**

```python
print(train["is_fraud"].mean() * 100, "% of train payments are fraud")
print(train["is_fraud"].sum(), "fraud payments in train")
print(test["is_fraud"].sum(), "fraud payments in test")
```

## **3. What information does each payment have? (column names and sample rows)**

```python
print(train.columns.tolist())
train.head(3)
```

## **4. What dates do the two files cover?**

```python
print(train["trans_date_trans_time"].min(), "to", train["trans_date_trans_time"].max())
print(test["trans_date_trans_time"].min(), "to", test["trans_date_trans_time"].max())
```

```python
print(train["cc_num"].nunique(), "different cards in train")
print(test["cc_num"].nunique(), "different cards in test")
print(len(set(test["cc_num"]) & set(train["cc_num"])), "test cards also appear in train")
print(train.groupby("cc_num").size().describe())
```
